```python
import sys
import joblib

sys.path.insert(0, '../..')  # Load Parent directory to import utils
from utils.constants import Configs
```

```python
for rep in ("ShapeBased", "FeatureBased"):
    oof = joblib.load(Configs.loso_results_path("DBSCAN", rep, "Pick"))["OOF_Movement"]
    for name, part in (("movID 1", oof[oof.movID == 1]), ("movID 3,5,7", oof[oof.movID != 1])):
        n = len(part)
        flagged = int(part.pred_is_outlier.sum())
        tp = int(((part.pred_is_outlier == 1) & (part.true_is_outlier == 1)).sum())
        print(f"{rep:13s} {name:12s} n={n:3d}  sinalizados={flagged:3d} ({100*flagged/n:.1f}%)  "
              f"anotados={int(part.true_is_outlier.sum())}  TP={tp}  FP={flagged - tp}")
```
